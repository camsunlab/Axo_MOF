```python
# Vacuum Gripper functions 

# Picks up the griper tool
axo.pickup_tool(gripper)
print("Picked Up Vacuum Gripper")

# get a lid from lid storage on slot 4, then moves to lot 3 and places a lid on top of the washing vials
gripper.pick_and_place(vacuum_location[4], # This places the lid onto the washing place
                       vacuum_location[3], 0.7, 3) # the washing station
print("Placed the lid on the plate") 

# pick up the lid and return it to the lid storage
gripper.grip(vacuum_location[3], 0.7, 3) # This picks up the lid from the washing
gripper.drop(vacuum_location[4], 3)
print("Return the lid from the plate") 

# Parks the tool
axo.park_tool()
print("Parked Vacuum Gripper")
```
